## Affichage d'une image à niveaux de gris

La conversion d'une image couleur en niveau de gris s'effectue en remplaçant chaque pixel de valeurs $(r,g,b)$ en un pixel d'intensité $x$ tel que
$x = 0.2125 \, r + 0.7154 \, g + 0.0721 \, b$.
Les coefficients ont été définis à partir de mesures psychovisuelles et de façon à garantir que $x\in[0,1]$ si $r,g,b\in[0,1]$.

* Convertissez l'image couleur en niveaux de gris (`skimage.color.rgb2gray`).

* Affichez l'image à niveau de gris $g$
  en ajoutant également la barre de couleur
  et les valeurs minimales et maximales de ses pixels :
  ```
  plt.imshow(g)
  plt.colorbar()
  print(g.min())
  print(g.max())
  ```
  

* Utilisez l'argument `cmap` de `imshow`
  pour modifier la palette des couleurs (_colormap_).
  Pour cela, consultez l'aide de la fonction :
  ```
  help(plt.imshow)
  ```

* Affichez le profil d'intensité de la colonne 260 :
  ```
  plt.plot(g[:,260])
  ```
  À quoi correpondent la droite décroissante et les deux pics vers le bas ?
