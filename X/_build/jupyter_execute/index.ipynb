# Introduction

Ce manuel regroupe les TP de Traitement d'images du département I2S de [Télécom Physique Strasbourg](https://www.telecom-physique.fr/).
Le manuel de cours est accessible sur [Basics of Image Processing](https://vincmazet.github.io/bip/).

Les TP sont des exercices de programmation en Python (une aide est disponible dans la page [](P:python)).
Des ressources complémentaires sont disponibles sur la page Moodle dédiée (accès réservé à mes étudiants).
