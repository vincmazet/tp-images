# Produit de convolution

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.ndimage as ndi
```

```python
def myimshow(ax, img, title):
    """Affichage perso"""
    
    # Paramètres d'affichage
    clr = '#0c0'
    vmin, vmax = -1, 4
    
    # Taille de l'image
    M, N = img.shape

    # Image
    ax.imshow(img, cmap="gray", interpolation='none', aspect='equal', vmin=vmin, vmax=vmax)

    # Ticks et grille
    ax.set_xticks(np.arange(0, N, 1))
    ax.set_yticks(np.arange(0, M, 1))
    ax.set_xticklabels(np.arange(0, N, 1))
    ax.set_yticklabels(np.arange(0, M, 1))
    ax.set_xticks(np.arange(-.5, N, 1), minor=True)
    ax.set_yticks(np.arange(-.5, M, 1), minor=True)
    ax.grid(which='minor', color=clr, linestyle='-', linewidth=1)
    ax.tick_params(which='minor', bottom=False, left=False)

    # Couleur des axes
    ax.spines['bottom'].set_color(clr)
    ax.spines['top'].set_color(clr) 
    ax.spines['right'].set_color(clr)
    ax.spines['left'].set_color(clr)

    # Intensités
    for m in range(M):
        for n in range(N):
            ax.text(n, m, img[m,n], va="center", ha="center", color=clr)

    # Titre
    ax.set_title(title)
```

```python
# Image
ligne = [1, 1, 1, 0, 0, 0]
g  = np.array([ligne, ligne])

# PSF
h1 = np.array([ [1, -1] ])
h2 = np.array([ [1], [-1] ])
h3 = np.array([ [1, 2, 1] ])

for h in [h1, h2, h3]:

    # Convolution
    f = ndi.convolve(g, h)

    # Affichage
    fig, axs = plt.subplots(1, 3, figsize=(12,1))
    myimshow(axs[0], g, "Originale")
    myimshow(axs[1], h, "PSF")
    myimshow(axs[2], f, "Résultat de la convolution")
    plt.show
```
